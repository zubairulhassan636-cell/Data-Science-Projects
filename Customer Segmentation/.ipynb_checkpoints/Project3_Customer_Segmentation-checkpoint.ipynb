# Project 3 — Unsupervised Learning (Customer Segmentation)
### DecodeLabs Data Science Internship | Batch 2026

**Goal:** Distance-based algorithms (K-Means) use karke unlabeled retail data mein hidden
mathematical customer groupings discover karna, PCA se dimensionality reduce karna, aur
resulting clusters ko actionable business "Personas" mein translate karna.

**Pipeline (IPO Architecture):**
1. **Scale** — Standardization (input)
2. **Compress** — PCA (process)
3. **Cluster** — K-Means (process)
4. **Translate** — Business Personas (output)

---
**Dataset:** `retail_customers.csv` — 1200 customers, 23 features (demographics, spending
behavior, engagement, category-wise spend) — bilkul unlabeled hai, koi target column nahi.


```python
# 1. Imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from kneed import KneeLocator

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (8, 5)

pd.set_option('display.max_columns', 30)
np.random.seed(42)
```

20+ numeric features mein correlations dekhna zaroori hai — high correlation wale

## Phase 1: Scale — Standardization
`StandardScaler` har feature ko mean=0, std=1 par map karta hai, taake har feature ki
"mathematical voting power" equal ho jaye — koi bhi single feature (jaise Income) apni
badi magnitude ki wajah se distance calculation ko hijack na kar sake.

$$z = \dfrac{x - \mu}{\sigma}$$

Gender ek categorical column hai isliye clustering ke numeric features mein include
nahi karenge (yeh sirf persona description ke liye baad mein use hoga).

```python
# Sirf numeric behavioral/demographic features clustering ke liye use karenge
feature_cols = [c for c in numeric_df.columns]
X = df[feature_cols].copy()

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Scaled feature matrix shape:", X_scaled.shape)
print("Mean (should be ~0):", X_scaled.mean(axis=0).round(2)[:5])
print("Std (should be ~1):", X_scaled.std(axis=0).round(2)[:5])
```

## Phase 2: Compress — PCA (Curse of Dimensionality)
23 features high-dimensional space banate hain jahan Euclidean distance meaningless ho
jati hai (sab points equidistant lagte hain). PCA orthogonal axes (Principal Components)
dhoondta hai jo maximum variance capture karte hain, aur hum data ko un axes par project
kar dete hain.

**95% Rule:** Hum utne components rakhte hain jitne cumulative explained variance ka
95% cover karein — baaki low-variance "noise" discard ho jata hai.

```python
# Full PCA fit karke explained variance dekhte hain
pca_full = PCA().fit(X_scaled)
cum_var = np.cumsum(pca_full.explained_variance_ratio_)

n_components_95 = np.argmax(cum_var >= 0.95) + 1
print(f"Number of components needed for 95% variance: {n_components_95}")

plt.figure(figsize=(9, 5))
plt.plot(range(1, len(cum_var)+1), cum_var, marker='o', color='#2C3E50')
plt.axhline(0.95, color='#E67E22', linestyle='--', label='95% Variance Threshold')
plt.axvline(n_components_95, color='#E67E22', linestyle=':')
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('The 95% Rule — Separating Signal from Noise')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Ab final PCA fit karte hain chosen number of components ke sath
pca = PCA(n_components=n_components_95, random_state=42)
X_pca = pca.fit_transform(X_scaled)

print("Reduced shape:", X_pca.shape)
print("Variance explained by each component:", pca.explained_variance_ratio_.round(3))
```

## Phase 3: Cluster — K-Means + Proving Optimal K
K-Means khud decide nahi karta ke kitne clusters hain — hum ne mathematically prove
karna hai. Do "diagnostic gatekeepers" use karenge:

1. **Elbow Method** — WCSS (Within-Cluster Sum of Squares) vs K plot, jahan curve ka
   "elbow" (maximum curvature point) optimal K batata hai.
2. **Silhouette Score** — cluster cohesion vs separation measure karta hai; jitna
   +1 ke qareeb utna behtar.

```python
wcss = []
sil_scores = []
K_range = range(2, 11)

for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_pca)
    wcss.append(km.inertia_)
    sil_scores.append(silhouette_score(X_pca, labels))

# Kneedle algorithm se elbow point automatically detect karte hain
knee = KneeLocator(list(K_range), wcss, curve='convex', direction='decreasing')
elbow_k = knee.knee
print("Elbow method suggests K =", elbow_k)
print("Best K by Silhouette Score =", list(K_range)[np.argmax(sil_scores)])
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(list(K_range), wcss, marker='o', color='#2C3E50')
if elbow_k:
    axes[0].axvline(elbow_k, color='#C0392B', linestyle='--', label=f'Elbow at K={elbow_k}')
axes[0].set_xlabel('Number of Clusters (K)')
axes[0].set_ylabel('WCSS')
axes[0].set_title('Diagnostic Gatekeeper 1: Elbow Method')
axes[0].legend()

axes[1].plot(list(K_range), sil_scores, marker='o', color='#8E44AD')
axes[1].set_xlabel('Number of Clusters (K)')
axes[1].set_ylabel('Silhouette Score')
axes[1].set_title('Diagnostic Gatekeeper 2: Silhouette Score')

plt.tight_layout()
plt.show()
```

### Final K Selection
Elbow method aur Silhouette score dono ko combine karke final K choose karte hain.
Hamare synthetic dataset mein humne 4 natural segments design kiye the, is liye
expected answer K=4 ke qareeb hona chahiye — is se pipeline ki correctness validate
hoti hai.

```python
optimal_k = elbow_k if elbow_k else list(K_range)[np.argmax(sil_scores)]
print(f"Final chosen K = {optimal_k}")

kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
cluster_labels = kmeans.fit_predict(X_pca)

df['Cluster'] = cluster_labels
final_silhouette = silhouette_score(X_pca, cluster_labels)
print(f"Final Silhouette Score for K={optimal_k}: {final_silhouette:.3f}")
df['Cluster'].value_counts().sort_index()
```

### Cluster Visualization (PCA Space)
Clusters ko pehle 2 principal components par plot karte hain taake visually confirm
ho sake ke separation kaisi hai.

```python
plt.figure(figsize=(9, 7))
palette = sns.color_palette('Set1', n_colors=optimal_k)
for c in range(optimal_k):
    mask = cluster_labels == c
    plt.scatter(X_pca[mask, 0], X_pca[mask, 1], s=25, alpha=0.6,
                color=palette[c], label=f'Cluster {c}')

centers_pca = kmeans.cluster_centers_
plt.scatter(centers_pca[:, 0], centers_pca[:, 1], s=250, c='black', marker='X',
            label='Centroids', edgecolors='white', linewidths=1.5)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title(f'Customer Segments in PCA Space (K={optimal_k})')
plt.legend()
plt.tight_layout()
plt.show()
```

## Phase 4: Translate — Reverse-Engineering the Centroids
K-Means clustering PCA-space mein hui thi, is liye centroids ke coordinates abstract
hain (marketing team ke liye meaningless). Hum inverse transforms (PCA → Scaler)
apply karke original human-readable units (Age, Income, Spending Score, etc.) mein
wapas laate hain.

$$C_{original} = \text{scaler.inverse\_transform}(\text{pca.inverse\_transform}(C_{pca}))$$

```python
centroids_original_scale = scaler.inverse_transform(
    pca.inverse_transform(kmeans.cluster_centers_)
)

centroid_df = pd.DataFrame(centroids_original_scale, columns=feature_cols)
centroid_df.insert(0, 'Cluster', range(optimal_k))
centroid_df.round(1)
```

## The Strategic Persona Matrix
Ab har cluster ke actual (reverse-engineered) averages dekh kar unhe business persona
naam aur recommended action assign karte hain — yeh step data science ko business
strategy mein translate karta hai.

```python
summary_cols = ['Age', 'Annual_Income', 'Spending_Score',
                 'Purchase_Frequency_Monthly', 'Online_Purchase_Ratio', 'Loyalty_Points']

persona_summary = df.groupby('Cluster')[summary_cols].mean().round(1)
persona_summary['Count'] = df['Cluster'].value_counts().sort_index()
persona_summary['Female_%'] = (df.groupby('Cluster')['Gender']
                                .apply(lambda s: (s == 'Female').mean() * 100).round(1))
persona_summary
```

```python
def assign_persona(row):
    """Income aur Spending Score ke combination se business persona assign karta hai."""
    high_income = row['Annual_Income'] >= persona_summary['Annual_Income'].median()
    high_spend = row['Spending_Score'] >= persona_summary['Spending_Score'].median()

    if high_income and high_spend:
        return "High-Value Trendsetters", "Exclusive perks, early access, experiential marketing."
    elif high_income and not high_spend:
        return "Affluent Conservatives", "High-touch support, warranties, loyalty programs."
    elif not high_income and high_spend:
        return "Budget-Conscious Explorers", "Influencer campaigns, flash sales, buy-now-pay-later."
    else:
        return "Conservative Minimizers", "Minimize spend, clear price-value, basic utility."

persona_names, actions = [], []
for _, row in persona_summary.iterrows():
    name, action = assign_persona(row)
    persona_names.append(name)
    actions.append(action)

persona_summary['Persona'] = persona_names
persona_summary['Recommended_Action'] = actions
persona_summary[['Persona', 'Age', 'Annual_Income', 'Spending_Score',
                  'Purchase_Frequency_Monthly', 'Count', 'Female_%', 'Recommended_Action']]
```

### Visual Persona Map
Income vs Spending Score par clusters plot karke persona quadrants dikhate hain —
yeh exactly wahi "Strategic Persona Matrix" concept hai jo slides mein tha.

```python
plt.figure(figsize=(9, 7))
for c in range(optimal_k):
    mask = df['Cluster'] == c
    plt.scatter(df.loc[mask, 'Annual_Income'], df.loc[mask, 'Spending_Score'],
                s=25, alpha=0.5, color=palette[c], label=f'Cluster {c}: {persona_names[c]}')

plt.axvline(df['Annual_Income'].median(), color='gray', linestyle='--', alpha=0.5)
plt.axhline(df['Spending_Score'].median(), color='gray', linestyle='--', alpha=0.5)
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.title('Customer Personas: Income vs Spending Behavior')
plt.legend(loc='best', fontsize=9)
plt.tight_layout()
plt.show()
```

## Save Results
Final labeled dataset (with Cluster + Persona) aur centroid summary ko CSV mein
export kar rahe hain — yeh business/marketing team ko share kiya ja sakta hai.

```python
cluster_to_persona = dict(zip(range(optimal_k), persona_names))
df['Persona'] = df['Cluster'].map(cluster_to_persona)

df.to_csv('customer_segments_labeled.csv', index=False)
persona_summary.to_csv('persona_summary.csv')

print("Saved: customer_segments_labeled.csv, persona_summary.csv")
df[['CustomerID', 'Age', 'Annual_Income', 'Spending_Score', 'Cluster', 'Persona']].head(10)
```

## Conclusion

Is project mein humne complete **IPO (Input-Process-Output)** unsupervised learning
pipeline implement kiya:

1. **Scale** — `StandardScaler` se 23 features ko equal mathematical voting power di.
2. **Compress** — `PCA` se dimensionality ko 95% variance retain karte hue kam kiya.
3. **Cluster** — `K-Means` ka optimal K, Elbow Method + Silhouette Score se mathematically
   prove kiya.
4. **Translate** — Cluster centroids ko inverse-transform karke real business
   **Personas** aur **actionable recommendations** banayi.

Yeh workflow real-world enterprise recommendation engines, targeted marketing campaigns,
aur churn-risk identification ke liye directly use ho sakta hai. Apna khud ka dataset
(retail_customers.csv ki jagah) daal kar yehi pipeline kisi bhi unlabeled tabular data
