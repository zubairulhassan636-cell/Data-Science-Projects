## 1. Install / import libraries

```python
# Run this cell once if packages are missing:
# %pip install pandas numpy matplotlib seaborn scikit-learn scipy nltk

import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk import word_tokenize, pos_tag
from nltk.corpus.reader import WordNetCorpusReader

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB, ComplementNB
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
from scipy.sparse import issparse

pd.set_option("display.max_colwidth", 120)
RANDOM_STATE = 42
```

## 2. Download required NLTK resources

```python
# These downloads are required for stop-word removal, WordNet lemmatization,
# and POS tagging. They are downloaded only to the local NLTK data directory.

resources = [
    "stopwords",
    "wordnet",
    "omw-1.4",
    "averaged_perceptron_tagger_eng"
]

for resource in resources:
    nltk.download(resource, quiet=True)

print("NLTK resources are ready.")
```

## 3. Load the dataset

```python
DATA_PATH = "product_reviews.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)
display(df.head(10))

print("\nClass distribution:")
display(df["sentiment"].value_counts())
```

## 4. Basic data validation and EDA

```python
print("Missing values:")
display(df.isna().sum())

print("\nDuplicate rows:", df.duplicated().sum())

df["word_count"] = df["review"].str.split().str.len()

print("\nText length summary:")
display(df["word_count"].describe())

# Sentiment distribution
plt.figure(figsize=(6, 4))
df["sentiment"].value_counts().plot(kind="bar")
plt.title("Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Review length distribution
plt.figure(figsize=(7, 4))
df.groupby("sentiment")["word_count"].plot(kind="hist", alpha=0.55, bins=15, legend=True)
plt.title("Review Length Distribution")
plt.xlabel("Number of Words")
plt.tight_layout()
plt.show()
```

## 5. Text preprocessing pipeline

The project specification requires a strict sequence:

**Raw text → normalization → tokenization → stop-word removal → POS tagging → lemmatization → cleaned text**

### Important engineering decision
Default English stop-word lists can remove words such as **not**, **no**, and **never**. Those words can reverse sentiment, so they are explicitly retained.


```python
lemmatizer = WordNetLemmatizer()

# Start with NLTK's English stop words, but preserve sentiment-critical negations.
stop_words = set(stopwords.words("english"))
negations_to_keep = {"no", "not", "nor", "never"}
stop_words = stop_words - negations_to_keep

def treebank_to_wordnet(tag):
    if tag.startswith("J"):
        return "a"  # adjective
    if tag.startswith("V"):
        return "v"  # verb
    if tag.startswith("N"):
        return "n"  # noun
    if tag.startswith("R"):
        return "r"  # adverb
    return "n"

def preprocess_text(text):
    # Normalize contractions so "isn't" becomes "is not".
    text = str(text).lower()
    text = re.sub(r"n't\b", " not", text)

    # Tokenization
    tokens = word_tokenize(text, preserve_line=True)

    # Keep alphabetic tokens and remove punctuation/numbers.
    tokens = [tok for tok in tokens if tok.isalpha()]

    # Stop-word removal while preserving negations.
    tokens = [tok for tok in tokens if tok not in stop_words]

    # POS-guided lemmatization.
    tagged_tokens = pos_tag(tokens)
    lemmas = [
        lemmatizer.lemmatize(token, treebank_to_wordnet(tag))
        for token, tag in tagged_tokens
    ]

    return " ".join(lemmas)

df["clean_text"] = df["review"].apply(preprocess_text)

display(df[["review", "clean_text", "sentiment"]].head(15))
```

### 5.1 Verify that negations are preserved

```python
negation_examples = df[df["review"].str.contains(r"\bnot\b|\bnever\b|\bno\b", case=False, regex=True)]
display(negation_examples[["review", "clean_text", "sentiment"]].head(10))

# A small unit-style check
assert "not" in preprocess_text("This product is not good.")
print("Negation-preservation check passed.")
```

## 6. Train/test split

A stratified split keeps the positive/negative class ratio approximately equal in both sets.


```python
X = df["clean_text"]
y = df["sentiment"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("\nTraining class distribution:")
display(y_train.value_counts(normalize=True))
```

## 7. TF-IDF vectorization

TF-IDF converts text into a numerical feature matrix.

- **TF (Term Frequency):** how frequently a term occurs in a document.
- **IDF (Inverse Document Frequency):** reduces the weight of terms that occur across many documents.
- **N-grams:** unigrams + bigrams help capture phrases such as **"not good"**.
- `max_features=10000` limits vocabulary size.
- `min_df=2` removes terms that occur in only one document.


```python
vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    max_features=10_000,
    min_df=2,
    sublinear_tf=True,
    strip_accents="unicode"
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("TF-IDF training shape:", X_train_tfidf.shape)
print("TF-IDF testing shape :", X_test_tfidf.shape)
print("Vocabulary size      :", len(vectorizer.vocabulary_))
print("Training matrix sparse:", issparse(X_train_tfidf))
print("Matrix format         :", type(X_train_tfidf).__name__)

# CSR is the standard sparse representation used by SciPy/scikit-learn.
X_train_csr = X_train_tfidf.tocsr()
X_test_csr = X_test_tfidf.tocsr()

print("CSR training matrix:", type(X_train_csr).__name__)
```

### 7.1 Inspect important TF-IDF features

```python
feature_names = np.array(vectorizer.get_feature_names_out())

# Average TF-IDF weight across training documents
mean_tfidf = np.asarray(X_train_csr.mean(axis=0)).ravel()
top_indices = mean_tfidf.argsort()[-20:][::-1]

top_features = pd.DataFrame({
    "feature": feature_names[top_indices],
    "mean_tfidf": mean_tfidf[top_indices]
})

display(top_features)
```

## 8. Train the Multinomial Naive Bayes classifier

Multinomial Naive Bayes is a natural baseline for text classification.

The `alpha` parameter applies **Laplace/additive smoothing**, preventing zero probabilities for unseen features.


```python
model = MultinomialNB(alpha=1.0)
model.fit(X_train_csr, y_train)

y_pred = model.predict(X_test_csr)

accuracy = accuracy_score(y_test, y_pred)

print(f"Test Accuracy: {accuracy:.4f} ({accuracy * 100:.2f}%)")
print("\nClassification Report:")
print(classification_report(y_test, y_pred, digits=4))
```

## 9. Confusion matrix

```python
cm = confusion_matrix(y_test, y_pred, labels=["negative", "positive"])

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Negative", "Positive"]
)
disp.plot()
plt.title("Multinomial Naive Bayes — Confusion Matrix")
plt.tight_layout()
plt.show()
```

## 10. Optional: compare MultinomialNB with ComplementNB

The project blueprint notes that `ComplementNB` can be useful when classes are imbalanced. Our dataset is balanced, so `MultinomialNB` is the main model, but we can still compare both classifiers.


```python
comparison_models = {
    "MultinomialNB": MultinomialNB(alpha=1.0),
    "ComplementNB": ComplementNB(alpha=1.0)
}

comparison_results = []

for name, clf in comparison_models.items():
    clf.fit(X_train_csr, y_train)
    pred = clf.predict(X_test_csr)
    comparison_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, pred)
    })

comparison_df = pd.DataFrame(comparison_results).sort_values("Accuracy", ascending=False)
display(comparison_df)
```

## 11. Inference: predict sentiment for new reviews

The final production-style function applies **the same preprocessing and TF-IDF vectorizer** used during training before making a prediction.


```python
def predict_sentiment(review, classifier=model):
    cleaned = preprocess_text(review)
    features = vectorizer.transform([cleaned]).tocsr()

    prediction = classifier.predict(features)[0]

    # MultinomialNB exposes class probabilities.
    probabilities = classifier.predict_proba(features)[0]
    probability_map = dict(zip(classifier.classes_, probabilities))

    return {
        "review": review,
        "cleaned_text": cleaned,
        "sentiment": prediction,
        "negative_probability": probability_map.get("negative", np.nan),
        "positive_probability": probability_map.get("positive", np.nan)
    }

new_reviews = [
    "The battery is excellent and the screen is bright.",
    "This product is not good and stopped working quickly.",
    "The headphones are comfortable and the sound is clear.",
    "The package arrived damaged and the device is unreliable."
]

predictions = pd.DataFrame([predict_sentiment(r) for r in new_reviews])
display(predictions)
```

## 12. Error analysis

Accuracy alone does not explain *why* predictions are wrong. Reviewing misclassified examples helps identify vocabulary overlap, ambiguous language, and dataset limitations.


```python
error_df = pd.DataFrame({
    "review": X_test.index.map(df["review"]),
    "actual": y_test.values,
    "predicted": y_pred
})

errors = error_df[error_df["actual"] != error_df["predicted"]]
print("Number of misclassified test reviews:", len(errors))
display(errors.head(20))
```
